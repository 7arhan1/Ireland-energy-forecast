Merge TSO + DSO Wind capacity files
Add Zone mapping (East / West)
Group & sum by Zone, Type, Connection_Type
Save both the merged raw file & the grouped zone table

```python
import pandas as pd

tso_file = '../data/processed/TSO_Wind_Capacity_Clean.csv'
dso_file = '../data/processed/DSO_Wind_Capacity_Clean.csv'
merged_output = '../data/processed/Wind_Capacity_Master.csv'
# regional_output = '../data/processed/Regional_Wind_Capacity.csv'


# Load TSO & DSO 
df_tso = pd.read_csv(tso_file)
df_dso = pd.read_csv(dso_file)


if 'Category' not in df_dso.columns:
    df_dso['Category'] = 'Onshore'


# Merge both
df_merged = pd.concat([df_tso, df_dso], ignore_index=True)


# Define Republic-only zones 
east_counties = [
    "Dublin", "Meath", "Wicklow", "Louth", "Kildare", "Offaly",
    "Laois", "Carlow", "Wexford", "Kilkenny", "Longford", "Westmeath"
]

west_counties = [
    "Galway", "Mayo", "Clare", "Sligo", "Leitrim", "Roscommon"
]

north_counties = [
    "Donegal", "Cavan", "Monaghan"
]

south_counties = [
    "Cork", "Kerry", "Waterford", "Tipperary", "Limerick"
]


# NI counties to filter out
ni_counties = ["Derry", "Tyrone", "Antrim", "Down", "Armagh", "Fermanagh"]


# Map County to Zone 
def county_to_zone(county):
    if county in ni_counties:
        return 'NI'
    elif county in east_counties:
        return 'East'
    elif county in west_counties:
        return 'West'
    elif county in north_counties:
        return 'North'
    elif county in south_counties:
        return 'South'
    else:
        return 'Other'



df_merged['Zone'] = df_merged['County'].apply(county_to_zone)


# Inspect any that ended up "Other" for QC
print("\n Counties mapped as 'Other':")
print(df_merged[df_merged['Zone'] == 'Other']['County'].unique())


# Save full Wind merged block
df_merged.to_csv(merged_output, index=False)
print(f" Wind capacity master saved to: {merged_output}")


# Filter out NI rows for regional grouping
df_regional = df_merged[df_merged['Zone'] != 'NI']


# Group by Zone, Type, Connection_Type & sum Installed MW
df_regional = (
    df_regional.groupby(['Zone', 'Type', 'Connection_Type'])
    .agg({'Installed_MW': 'sum'})
    .reset_index()
)

# #  Save Regional Wind Capacity table
# df_regional.to_csv(regional_output, index=False)
# print(f" Regional wind capacity saved to: {regional_output}")
```

```python
import pandas as pd


INPUT_FILE = r"C:\Projects\GitHub\Ireland-energy-forecast\data\processed\Solar_Capacity_Master.csv"
OUTPUT_FILE = r"C:\Projects\GitHub\Ireland-energy-forecast\data\processed\Aggregated_Solar_Capacity_By_County.csv"

# Latitude and Longitude mapping 
county_coords = {
    "Wexford": (52.3369, -6.4629),
    "Meath": (53.6055, -6.6564),
    "Cork": (51.8985, -8.4756),
    "Dublin": (53.3331, -6.2489),
    "Kildare": (53.1589, -6.9093),
    "Galway": (53.2719, -9.0489),
    "Waterford": (52.2593, -7.1101),
    "Offaly": (53.2734, -7.4903),
    "Tipperary": (52.4738, -8.1619),
    "Limerick": (52.6647, -8.6231),
    "Louth": (53.9500, -6.5400),
    "Kerry": (52.1545, -9.5669),
    "Kilkenny": (52.6541, -7.2448)
}

df = pd.read_csv(INPUT_FILE)

# Clean county names, remove extra spaces, capitalize
df["County"] = df["County"].str.strip().str.title()

# Aggregate Installed_MW by County & Connection_Type 
agg_df = df.groupby(["County", "Connection_Type"], as_index=False)["Installed_MW"].sum()

# Pivot to get TSO and DSO as separate columns
pivot_df = agg_df.pivot(index="County", columns="Connection_Type", values="Installed_MW").fillna(0)

# Ensuring both TSO and DSO exist as columns
if "TSO" not in pivot_df.columns:
    pivot_df["TSO"] = 0
if "DSO" not in pivot_df.columns:
    pivot_df["DSO"] = 0

# Calculate Total_MW 
pivot_df["Total_MW"] = pivot_df["TSO"] + pivot_df["DSO"]

# Adding coordinates 
pivot_df["Latitude"] = pivot_df.index.map(lambda x: county_coords.get(x, (None, None))[0])
pivot_df["Longitude"] = pivot_df.index.map(lambda x: county_coords.get(x, (None, None))[1])

# Keeping only required columns and reorder
final_df = pivot_df.reset_index()[["County", "Longitude", "Latitude", "Total_MW"]]

final_df.to_csv(OUTPUT_FILE, index=False)

print(f" Aggregated solar capacity saved to:\n{OUTPUT_FILE}")
print(final_df.head())
```
